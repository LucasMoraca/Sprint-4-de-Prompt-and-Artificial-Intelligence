## 1. Preparar o projeto
Se pedir, selecione o arquivo **sprint4_avaliacao.zip**.

```python
import os, sys, glob, zipfile, subprocess

def achar_projeto():
    for base in [os.getcwd(), "/content"]:
        for g in [os.path.join(base, "golden_dataset.json"), os.path.join(base, "*", "golden_dataset.json"),
                  os.path.join(base, "*", "*", "golden_dataset.json")]:
            achados = glob.glob(g)
            if achados:
                return os.path.dirname(achados[0])
    return None

pasta = achar_projeto()
if pasta is None:
    from google.colab import files
    up = files.upload()                      # selecione o sprint4_avaliacao.zip
    for nome in up:
        if nome.endswith(".zip"):
            zipfile.ZipFile(nome).extractall("/content")
    pasta = achar_projeto()
assert pasta, "Não encontrei golden_dataset.json. Suba o sprint4_avaliacao.zip."
os.chdir(pasta)
if pasta not in sys.path:
    sys.path.insert(0, pasta)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"])
print("Projeto em:", os.getcwd())
print("Arquivos:", sorted(os.listdir(".")))
```
